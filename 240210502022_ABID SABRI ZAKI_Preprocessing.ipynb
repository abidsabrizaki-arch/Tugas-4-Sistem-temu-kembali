```python
# Nama: Abid Sabri Zaki
# NIM : 240210502022

import re
import pandas as pd

from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
```

```python
data = {
    'Dokumen': [
        "Keamanan siber sangat penting untuk melindungi data pribadi dan sistem komputer dari serangan berbahaya.",

        "Internet of Things memungkinkan berbagai perangkat terhubung ke internet dan saling bertukar data secara otomatis.",

        "Perpustakaan digital menyediakan berbagai koleksi buku elektronik yang dapat diakses mahasiswa melalui jaringan internet.",

        "Komputasi awan menyediakan layanan penyimpanan dan pengolahan data melalui jaringan internet tanpa harus menggunakan perangkat keras sendiri.",

        "E-Commerce memudahkan masyarakat melakukan transaksi jual beli secara online dengan menggunakan berbagai metode pembayaran digital."
    ]
}

df = pd.DataFrame(data)

df
```

```python
# Membuat stemmer
stemmer_factory = StemmerFactory()
stemmer = stemmer_factory.create_stemmer()

# Membuat stopword remover
stopword_factory = StopWordRemoverFactory()
stopword_remover = stopword_factory.create_stop_word_remover()
```

```python
def preprocess_text(text):
    # 1. Case Folding
    text = text.lower()

    # 2. Membersihkan angka
    text = re.sub(r'\d+', '', text)

    # 3. Membersihkan tanda baca dan karakter khusus
    text = re.sub(r'[^a-z\s]', '', text)

    # 4. Menghapus spasi berlebih
    text = re.sub(r'\s+', ' ', text).strip()

    # 5. Tokenisasi
    tokens = text.split()

    # 6. Stopword Removal
    text = ' '.join(tokens)
    text = stopword_remover.remove(text)

    # 7. Tokenisasi kembali
    tokens = text.split()

    # 8. Stemming
    tokens = [stemmer.stem(token) for token in tokens]

    return tokens
```

```python
df['Hasil_Preprocessing'] = df['Dokumen'].apply(preprocess_text)

df
```

```python
for i in range(2):
    print("Dokumen", i + 1)
    print("Sebelum preprocessing:")
    print(df.loc[i, 'Dokumen'])

    print("\nSesudah preprocessing:")
    print(df.loc[i, 'Hasil_Preprocessing'])

    print("=" * 70)
```

```python
# Menghitung jumlah token sebelum preprocessing
df['Token_Sebelum'] = df['Dokumen'].apply(
    lambda x: len(x.lower().split())
)

# Menghitung jumlah token sesudah preprocessing
df['Token_Sesudah'] = df['Hasil_Preprocessing'].apply(
    lambda x: len(x)
)

# Menghitung persentase pengurangan token
df['Persentase_Pengurangan'] = (
    (df['Token_Sebelum'] - df['Token_Sesudah'])
    / df['Token_Sebelum']
) * 100

# Membulatkan menjadi 2 angka desimal
df['Persentase_Pengurangan'] = df['Persentase_Pengurangan'].round(2)

# Menampilkan hasil
df[['Token_Sebelum', 'Token_Sesudah', 'Persentase_Pengurangan']]
```

```python
hasil_analisis = pd.DataFrame({
    'Dokumen': range(1, len(df) + 1),
    'Token Sebelum': df['Token_Sebelum'],
    'Token Sesudah': df['Token_Sesudah'],
    'Pengurangan (%)': df['Persentase_Pengurangan']
})

hasil_analisis
```

```python
for i, row in df.iterrows():
    print(f"Dokumen {i + 1}:")
    print("Asli       :", row['Dokumen'])
    print("Preprocessing:", row['Hasil_Preprocessing'])
    print()
```
